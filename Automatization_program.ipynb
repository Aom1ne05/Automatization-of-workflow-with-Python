# **Automatization program**

"You need to help the secretary automate their work. To do this, you need to write a program that will execute user commands based on the stored data.

The initial data has the following structure:"

````python
# 1. List of all documents:

documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'John Ray'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Olivia Cust'},
    {'type': 'insurance', 'number': '10006', 'name': 'Stan Smith'}
]


# 2. List of shelves where the documents are stored. If a document exists in documents, it must also be in directories.

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}
````

## **Task 1**

### **Item 1:** The user can find out the owner of a document by its number using the command `"Find"`.

Examples:

Enter command:
`Find`

Enter document number:
  `10006` <br>
Result:
  `Document owner: Stan Smith`

Enter command:
`Find`

Enter document number:
  `12345` <br>
Result:
  `Document not found in the database`

````python
def owner_of_doc(doc):
  number_of_document = input('Enter document number: ')
  for element in doc:
    if number_of_document == element['number']:
      print(f"Document owner: {element['name']}")
      break
  else:
    print('Document not found in the database')
````

### **Item 1.2.** The user can find out which shelf a document is stored on by its number using the command `"Shelf"`.

Examples:

Enter command:
`Shelf`

Enter document number:
`10006 `<br>
Result:
`The document is stored on shelf: 2`

Enter command:
`Shelf`

Enter document number:
`12345` <br>
Result:
`Document not found in the database`

````python
def shelf_of_dir(dir):
  number_of_document = input('Enter document number: ')
  for key,values in dir.items():
    if number_of_document in values:
      print(f"The document is stored on shelf: {key}")
      break
  else:
    print('Document not found in the database')
````

### **Item 1.3.** The user can view complete information on all documents using the command `"Info"`.

Example:

Enter command:
`Info` <br>

Result:
```
№: 2207 876234, Type: passport, Owner: John Ray, Storage Shelf: 1  
№: 11-2, Type: invoice, Owner: Olivia Cust, Storage Shelf: 1
№: 10006, Type: insurance, Owner: Stan Smith, Storage Shelf: 2
```

````python
def all_info(doc, dir):
    for element in doc:
        for key, values in dir.items():
            if element['number'] in values:
                print(f"№: {element['number']}, Type: {element['type']}, Owner: {element['name']}, Storage Shelf: {key}")
````

### **Item 1.4.** The user can add a new shelf using the command `"Add Shelf"`.

Examples:

Enter command:
`Add Shelf`

Enter shelf number:
`10` <br>

Result: `Shelf added. Current list of shelves: 1, 2, 3, 10.`

<br>Enter command:
`Add Shelf`

Enter shelf number:
`1`<br>

Result: `This shelf already exists. Current list of shelves: 1, 2, 3.`

````python
def add_shelf(dir):
  number_of_document = input('Enter shelf number: ')
  if number_of_document not in dir:
    dir.setdefault(number_of_document, [])
    print(f"Shelf added. Current list of shelves: {', '.join(list(dir.keys()))}")
  else:
    print(f"This shelf already exists. Current list of shelves: {', '.join(list(dir.keys()))}")
````

### **Item 1.5.** The user can delete an existing shelf from the data using the command `"Delete Shelf"` (only if it is empty).

Examples:

Enter command:
`Delete Shelf`

Enter shelf number:
`3`<br>
Result: `Shelf deleted. Current list of shelves: 1, 2.`

Enter command:
`Delete Shelf`

Enter shelf number:
`1`<br>
Result: `The shelf contains documents, remove them before deleting the shelf. Current list of shelves: 1, 2, 3.`

Enter command:
`Delete Shelf`

Enter shelf number:
`4`<br>
Result: `This shelf does not exist. Current list of shelves: 1, 2, 3.`

````python
def del_shelf(dir):
    number_of_document = input('Enter shelf number: ')
    if number_of_document in dir and len(dir[number_of_document]) == 0:
        del dir[number_of_document]
        print(f"Shelf deleted. Current list of shelves: {', '.join(list(dir.keys()))}")
    elif number_of_document in dir and len(dir[number_of_document]) != 0:
        print(f"The shelf contains documents, remove them before deleting the shelf. Current list of shelves: {', '.join(list(dir.keys()))}")
    else:
        print(f"This shelf does not exist. Current list of shelves: {', '.join(list(dir.keys()))}")
````

````python
def add_doc(doc, dir):
    number_of_document = input('Enter document number: ')
    type_of_document = input('Enter document type: ')
    name_of_user = (input('Enter document owner: ')).title()
    shelf_number = input('Enter storage shelf: ')
    if shelf_number in dir:
        dir[shelf_number].append(number_of_document)
        doc.append({'type' : type_of_document,'number' : number_of_document,'name' : name_of_user})
        print("Document added. Current list of documents: ")
        all_info(doc, dir)
    else:
        print("This shelf does not exist. Add a shelf with the command ads. \nCurrent list of documents: ")
        all_info(doc, dir)
````

### **Item 2.2.** The user can delete a document from the data using the command `"Delete Document"`.

Examples:

Enter command:
`Delete Document`

Enter document number:
`10006` <br>

Result:
```
Document deleted.
Current list of documents:
№: 2207 876234, Type: passport, Owner: John Ray, Storage Shelf: 1  
№: 11-2, Type: invoice, Owner: Olivia Cust, Storage Shelf: 1
```

Enter command:
`Delete Document`

Enter document number:
`123456` <br>

Result:
```
Document not found in the database.
Current list of documents:
№: 2207 876234, Type: passport, Owner: John Ray, Storage Shelf: 1  
№: 11-2, Type: invoice, Owner: Olivia Cust, Storage Shelf: 1
№: 10006, Type: insurance, Owner: Stan Smith, Storage Shelf: 2
```

````python
def del_doc(doc,dir):
    number_of_document = input('Enter document number ')
    res = False
    for x in range(len(doc)):
        if number_of_document == doc[x]['number']:
            res = True
            del doc[x]
            for values in dir.values():
                for element in values:
                    if number_of_document in values:
                        values.pop(values.index(element))
            break
    if res:
        print(f"Document deleted. \nCurrent list of documents: ")
        all_info(doc, dir)
    else:
        print(f"Document not found in the database. \nCurrent list of documents: ")
        all_info(doc, dir)
````

```
№: 11-2, Type: invoice, Owner: Olivia Cust, Storage Shelf: 1
№: 10006, Type: insurance, Owner: Stan Smith, Storage Shelf: 2
```

````python
def move_doc(doc, dir):
    number_of_document = input('Enter document number: ')
    shelf_number = input('Enter shelf number: ')
    res = False
    res_shelf = False
    for values in dir.values():
        for element in values:
            if number_of_document in values and shelf_number in dir:
                res = True
                dir[shelf_number].append(values.pop(values.index(number_of_document)))
            elif shelf_number not in dir:
                res_shelf = True

    if res:
        print("Document moved. Current list of documents: ")
        all_info(doc, dir)
    elif res_shelf:
        print(f"This shelf does not exist. \nCurrent list of shelves: {', '.join(list(dir.keys()))}")
    else:
        print(f"Document not found in the database. \nCurrent list of documents: ")
        all_info(doc, dir)
````

### **Program**

````python
def main(doc, dir):
  print("Hint: To find out what commands are available, enter 'Commands'")
  while True:
    command = (input('Enter command: ')).title()
    match command:
      case 'Commands':
        print(f"List of commands: \n1. The user can find out the owner of a document by its number using the command 'Find'. \
                                  \n2. The user can find out which shelf a document is stored on by its number using the command 'Shelf'. \
                                  \n3. The user can view complete information on all documents using the command 'Info'. \
                                  \n4. The user can add a new shelf using the command 'Add Shelf'. \
                                  \n5. The user can delete an existing shelf from the data using the command 'Delete Shelf' (only if it is empty). \
                                  \n6. The user can add a new document to the data using the command 'Add Document'. \
                                  \n7. The user can delete a document from the data using the command 'Delete Document'. \
                                  \n8. The user can move a document from one shelf to another using the command 'Move Document'. \
                                  \n9. The user can quit from program using the command 'Quit'. ")
      case 'Find':
        owner_of_doc(doc)
      case 'Shelf':
        shelf_of_dir(dir)
      case 'Info':
        all_info(doc, dir)
      case 'Add Shelf':
        add_shelf(dir)
      case 'Delete Shelf':
        del_shelf(dir)
      case 'Add Document':
        add_doc(doc, dir)
      case 'Delete Document':
        del_doc(doc, dir)
      case 'Move Document':
        move_doc(doc, dir)
      case 'Quit':
        print('Quit')
        break
      case _:
        print('Invalid command')
````

````python
main(documents, directories)
````
